## 📘 1. Introducción teórica

Una función `lambda` es una función anónima, pequeña y de una sola línea.

Tiene la forma general:

```python
lambda argumentos: expresión

Es equivalente a una función definida con def, pero más breve y solo para funciones simples.

#### 📌 Usos

```markdown
## ✅ 2. ¿Cuándo usar `lambda`?

- Cuando necesitamos funciones simples, de una sola expresión.
- Para pasarlas como argumentos a funciones como `map`, `filter`, `sorted`, etc.
- Para evitar definir funciones completas con `def` que se usan una sola vez.

### 🚫 No se recomienda cuando:
- La función es compleja o tiene múltiples líneas.
- Queremos añadir documentación o manejo de errores.

````python
# Ejemplo 1: Doble de un número
doble = lambda x: x * 2
print(doble(4))
````

````python
# Ejemplo 2: Suma de dos números
suma = lambda x, y: x + y
print(suma(3, 5)) 
````

````python
# Ejemplo 3: Obtener el último carácter de una palabra
ultimo = lambda s: s[-1]
print(ultimo("funciones lambda en python"))  # n
````

````python
# Ejemplo 4: Ordenar una lista de tuplas por el primer valor
pares = [(1, 3), (4, 1), (2, 2)]
ordenado = sorted(pares, key=lambda x: x[0])
print(ordenado)  

# Ejemplo 4: Ordenar una lista de tuplas por el segundo valor
pares = [(1, 3), (4, 1), (2, 2)]
ordenado = sorted(pares, key=lambda x: x[1])
print(ordenado)  
````

````python
# Ejemplo 5: Filtrar (por modulo) una lista de números pares con filter
numeros = [1, 2, 3, 4, 5, 6]
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(pares)  
````

````python
# Ejemplo 6: Elevar al cuadrado con map
numeros = [1, 2, 3, 4]
cuadrados = list(map(lambda x: x**2, numeros))
print(cuadrados)  
````

````python
# Ejemplo 7: Verificar si una palabra contiene una letra específica
contiene_a = lambda palabra: 'a' in palabra
print(contiene_a("hola"))   
print(contiene_a("hello"))  
````

````python
# Ejemplo 8: Devolver el mayor de dos números
mayor = lambda x, y: x if x > y else y
print(mayor(5, 8))  
````

````python
# Ejemplo 9: Ordenar una lista de strings por su longitud
palabras = ["las", "clases", "de", "IBD", "son", "las", "mejores"]
ordenadas = sorted(palabras, key=lambda x: len(x))
print(ordenadas)  # ['es', 'genial', 'python']
````

````python
# Ejemplo 10: Calcular el máximo de tres números
mayor = lambda a, b, c: max(a, b, c)
print(mayor(10, 20, 15))  # Salida: 20
````
